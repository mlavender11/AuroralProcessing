```python
import numpy as np
import cv2
from scipy.signal import wiener
import h5py
from auroral_processing.utils import norm as norm_utils
import matplotlib.pyplot as plt
```

```python
fn = '2013-03-27-CamSer7196_015400-015600.h5'
f = h5py.File(fn, 'r')
imgs = f['rawimg']
ut = f['ut1_unix']
```

```python
img = imgs[100]
norm = norm_utils.compute_norm(imgs, ut, 1)
```

```python
plt.imshow(img, 'gray', norm=norm) 
```

```python
filtered = wiener(img, 5)
plt.imshow(filtered, 'gray', norm=norm)
```
